# **Histogram**

# Draw a histogram of the Age column. What does the shape of the distribution tell you about the ages in the dataset?

```python

# Draw a histogram of the Age column
plt.hist(df["Age"], bins=10, color='skyblue', edgecolor='black')

# Add labels and title
plt.title("Histogram of Age Distribution")
plt.xlabel("Age (years)")
plt.ylabel("Frequency")

# Show the plot
plt.show()
```

# Plot a histogram for LungCap. Is the data mostly concentrated at lower or higher lung capacities?

```python
# Plot a histogram for LungCap
plt.hist(df["LungCap"], bins=10, color='lightgreen', edgecolor='black')

# Add title and labels
plt.title("Histogram of Lung Capacity")
plt.xlabel("Lung Capacity (Liters)")
plt.ylabel("Frequency")

# Display the plot
plt.show()
```

# Create a histogram of Height. What is the most common height range in the dataset?

```python
# Plot histogram for Height
plt.hist(df["Height"], bins=10, color='lightcoral', edgecolor='black')

# Add title and labels
plt.title("Histogram of Height Distribution")
plt.xlabel("Height (inches)")
plt.ylabel("Frequency")

# Show the plot
plt.show()
```

# Create a histogram for Height grouped by Gender. Are males generally taller than females?

```python
# Plot histograms of Height grouped by Gender
plt.hist(df[df["Gender"] == "male"]["Height"], bins=10, alpha=0.7, label="Male", color='skyblue', edgecolor='black')
plt.hist(df[df["Gender"] == "female"]["Height"], bins=10, alpha=0.7, label="Female", color='lightpink', edgecolor='black')

# Add title and labels
plt.title("Histogram of Height Grouped by Gender")
plt.xlabel("Height (inches)")
plt.ylabel("Frequency")

# Add legend
plt.legend()

# Show the plot
plt.show()
```

# **BoxPlot**

# Q1: Draw a boxplot of LungCap. What does the boxplot tell you about the spread of lung capacity values?

```python
# Draw the boxplot for Lung Capacity
plt.figure(figsize=(6, 4))
sns.boxplot(y=df['LungCap'], color='skyblue')
plt.title('Boxplot of Lung Capacity')
plt.ylabel('Lung Capacity (L)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

# Q2: Use a boxplot to compare LungCap for smokers (Smoke = yes) and non-smokers (Smoke = no). Do smokers tend to have higher or lower lung capacity?

```python
# Draw boxplot for Lung Capacity grouped by Smoke status
plt.figure(figsize=(6, 4))
sns.boxplot(x='Smoke', y='LungCap', data=df, palette='pastel')
plt.title('Comparison of Lung Capacity: Smokers vs Non-Smokers')
plt.xlabel('Smoking Status')
plt.ylabel('Lung Capacity (L)')
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()
```

# **Coefficient** **of Variation**

# Calculate the CV of LungCap for participants born by Caesarean versus those who were not

```python


# Function to calculate Coefficient of Variation
def coefficient_of_variation(series):
    return (series.std() / series.mean()) * 100

# CV for Caesarean = yes
cv_caesarean_yes = coefficient_of_variation(df[df['Caesarean'] == 'yes']['LungCap'])

# CV for Caesarean = no
cv_caesarean_no = coefficient_of_variation(df[df['Caesarean'] == 'no']['LungCap'])

print("CV of LungCap (Caesarean = yes):", round(cv_caesarean_yes, 2), "%")
print("CV of LungCap (Caesarean = no):", round(cv_caesarean_no, 2), "%")
```

# For individuals aged below 12, what is the coefficient of variation of LungCap?

```python
# Filter individuals aged below 12
below_12 = df[df['Age'] < 12]

# Calculate Coefficient of Variation for LungCap
cv_below_12 = (below_12['LungCap'].std() / below_12['LungCap'].mean()) * 100

print(" (CV) of LungCap for individuals aged below 12:", round(cv_below_12, 2), "%")
```
